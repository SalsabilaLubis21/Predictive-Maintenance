###**PREPROCESSING**

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

# 1. Feature Engineering
telemetry_df['datetime'] = pd.to_datetime(telemetry_df['datetime'])
features = ['volt', 'rotate', 'pressure', 'vibration']

# Calculate Mean and Std Dev per 24 hours as basic features
telemetry_mean_24h = telemetry_df.groupby('machineID').rolling(window=24, on='datetime')[features].mean().reset_index()
telemetry_sd_24h = telemetry_df.groupby('machineID').rolling(window=24, on='datetime')[features].std().reset_index()

# Rename the columns
telemetry_mean_24h.columns = ['machineID', 'datetime'] + [f'{f}_mean_24h' for f in features]
telemetry_sd_24h.columns = ['machineID', 'datetime'] + [f'{f}_sd_24h' for f in features]

# Merge telemetry features
telem_feat = pd.merge(telemetry_mean_24h, telemetry_sd_24h, on=['machineID', 'datetime'])

# 2. Encoding Categorical Features (machines_df)
machines_encoded = pd.get_dummies(machines_df, columns=['model'])


final_feat = pd.merge(telem_feat, machines_encoded, on='machineID')

# 3. Labeling
failures_df['datetime'] = pd.to_datetime(failures_df['datetime'])
final_data = pd.merge(final_feat, failures_df, on=['datetime', 'machineID'], how='left')
final_data['failure'] = final_data['failure'].fillna('none')
final_data['target'] = (final_data['failure'] != 'none').astype(int)

# Drop NaN because of rolling window
final_data = final_data.dropna()

# 4. Split Data (Time-based split)
final_data = final_data.sort_values(by='datetime')

X = final_data.drop(['datetime', 'machineID', 'failure', 'target'], axis=1)
y = final_data['target']

train_idx = int(len(final_data) * 0.7)
val_idx = int(len(final_data) * 0.85)

X_train, y_train = X.iloc[:train_idx], y.iloc[:train_idx]
X_val, y_val = X.iloc[train_idx:val_idx], y.iloc[train_idx:val_idx]
X_test, y_test = X.iloc[val_idx:], y.iloc[val_idx:]

print(f"Shape Training  : {X_train.shape}")
print(f"Shape Validation: {X_val.shape}")
print(f"Shape Testing   : {X_test.shape}")
print("\nBaseline data ready with basic feature engineering.")
```

###**ADVANCED MODEL (USING XBOOST)**

```python
import numpy as np
import pandas as pd
import xgboost as xgb
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, average_precision_score, precision_recall_curve




# 1. Feature Engineering

telemetry_df['datetime'] = pd.to_datetime(telemetry_df['datetime'])
errors_df['datetime']    = pd.to_datetime(errors_df['datetime'])
maint_df['datetime']     = pd.to_datetime(maint_df['datetime'])
failures_df['datetime']  = pd.to_datetime(failures_df['datetime'])

features = ['volt', 'rotate', 'pressure', 'vibration']


# A. Rolling 24h Mean & Std
telemetry_mean_24h = (telemetry_df.groupby('machineID')
                       .rolling(window=24, on='datetime')[features]
                       .mean().reset_index())

# taking feature values ​​from the previous period for each machine
telemetry_mean_24h[features] = telemetry_mean_24h.groupby('machineID')[features].shift(1)

telemetry_sd_24h = (telemetry_df.groupby('machineID')
                     .rolling(window=24, on='datetime')[features]
                     .std().reset_index())

# Using the standard deviation value from the previous 24 hours for each machine
telemetry_sd_24h[features] = telemetry_sd_24h.groupby('machineID')[features].shift(1)

telemetry_mean_24h.columns = ['machineID', 'datetime'] + [f'{f}_mean_24h' for f in features]
telemetry_sd_24h.columns   = ['machineID', 'datetime'] + [f'{f}_sd_24h' for f in features]
telemetry_feat = pd.merge(telemetry_mean_24h, telemetry_sd_24h, on=['machineID', 'datetime'])

# Rolling 24h Min & Max
telemetry_min_24h = (telemetry_df.groupby('machineID')
                      .rolling(window=24, on='datetime')[features]
                      .min().reset_index())

telemetry_min_24h[features] = telemetry_min_24h.groupby('machineID')[features].shift(1)

telemetry_max_24h = (telemetry_df.groupby('machineID')
                      .rolling(window=24, on='datetime')[features]
                      .max().reset_index())

telemetry_max_24h[features] = telemetry_max_24h.groupby('machineID')[features].shift(1)

telemetry_min_24h.columns = ['machineID', 'datetime'] + [f'{f}_min_24h' for f in features]
telemetry_max_24h.columns = ['machineID', 'datetime'] + [f'{f}_max_24h' for f in features]

# Rate of change
telemetry_diff = telemetry_df.copy()
for f in features:
    telemetry_diff[f'{f}_diff'] = telemetry_df.groupby('machineID')[f].diff()


tele_full = telemetry_feat.merge(telemetry_min_24h, on=['machineID', 'datetime'], how='left')
tele_full = tele_full.merge(telemetry_max_24h, on=['machineID', 'datetime'], how='left')
tele_full = tele_full.merge(
    telemetry_diff[['machineID', 'datetime', 'volt_diff', 'rotate_diff', 'pressure_diff', 'vibration_diff']],
    on=['machineID', 'datetime'], how='left'
)


print("Calculating the rolling count error feature...")

error_dummies = pd.get_dummies(errors_df, columns=['errorID'])
error_cols = [c for c in error_dummies.columns if c.startswith('errorID_')]

error_timeline = pd.merge(telemetry_df[['datetime', 'machineID']], error_dummies, on=['datetime', 'machineID'], how='left').fillna(0)

# Rolling count 24h & 72h with shift per machineID
error_count_24h = (error_timeline.groupby('machineID')
                   .rolling(window=24, on='datetime')[error_cols]
                   .sum().reset_index())

error_count_24h[error_cols] = error_count_24h.groupby('machineID')[error_cols].shift(1)
error_count_24h.columns = ['machineID', 'datetime'] + [f'{c}_count_24h' for c in error_cols]

# Calculating the number of errors for each machine over the last 72 hours.
error_count_72h = (error_timeline.groupby('machineID')
                   .rolling(window=72, on='datetime')[error_cols]
                   .sum().reset_index())

error_count_72h[error_cols] = error_count_72h.groupby('machineID')[error_cols].shift(1)
error_count_72h.columns = ['machineID', 'datetime'] + [f'{c}_count_72h' for c in error_cols]

error_feat = pd.merge(error_count_24h, error_count_72h, on=['machineID', 'datetime'])


print("Calculating the number of days since the last maintenance for each component...")

maint_dummies = pd.get_dummies(maint_df, columns=['comp'])
comp_cols = [c for c in maint_dummies.columns if c.startswith('comp_')]

comp_df = pd.merge(telemetry_df[['datetime', 'machineID']], maint_dummies, on=['datetime', 'machineID'], how='left')

for c in comp_cols:
   # Stores the time the error occurred, or NaT if no error occurred.
    comp_df[f'{c}_last'] = np.where(comp_df[c] == 1, comp_df['datetime'], pd.NaT)


    comp_df[f'{c}_last'] = comp_df.groupby('machineID')[f'{c}_last'].shift(1)
    comp_df[f'{c}_last'] = comp_df.groupby('machineID')[f'{c}_last'].ffill()

    # conversion to datetime64
    comp_df[f'{c}_last'] = pd.to_datetime(comp_df[f'{c}_last'])

    # Calculate the number of hours since the last maintenance.
    comp_df[f'hours_since_{c}'] = (comp_df['datetime'] - comp_df[f'{c}_last']).dt.total_seconds() / 3600.0
    comp_df[f'hours_since_{c}'] = comp_df[f'hours_since_{c}'].fillna(1000)

maint_feat_cols = [f'hours_since_{c}' for c in comp_cols]
maint_feat = comp_df[['machineID', 'datetime'] + maint_feat_cols]


# MERGE ALL FEATURES & METADATA MESIN

print("Combining the entire dataset...")

machines_encoded = pd.get_dummies(machines_df, columns=['model'])

df_final = tele_full.merge(error_feat, on=['machineID', 'datetime'], how='left')
df_final = df_final.merge(maint_feat, on=['machineID', 'datetime'], how='left')
df_final = df_final.merge(machines_encoded, on='machineID', how='left')


# LABELING TARGET

print("Creating future prediction target ...")

failures_df['failure_event'] = 1
df_final = pd.merge(df_final, failures_df[['datetime', 'machineID', 'failure_event']], on=['datetime', 'machineID'], how='left')
df_final['failure_event'] = df_final['failure_event'].fillna(0)

df_final = df_final.sort_values(by=['machineID', 'datetime']).reset_index(drop=True)


df_final['future_failure'] = df_final.groupby('machineID')['failure_event'].shift(-1)

# Take the upcoming 24-hour window.
indexer = pd.api.indexers.FixedForwardWindowIndexer(window_size=24)
df_final['target'] = (df_final.groupby('machineID')['future_failure']
                      .rolling(window=indexer, min_periods=1)
                      .max()
                      .reset_index(drop=True))

df_final['target'] = df_final['target'].fillna(0)

# Reorder by global time
df_final = df_final.sort_values(by='datetime').reset_index(drop=True)

# Dropping Extra Columns & Handling NaN
feature_cols = [c for c in df_final.columns if c not in ['datetime', 'machineID', 'failure_event', 'future_failure', 'target']]
df_final = df_final.dropna(subset=feature_cols)


# TIME-BASED SPLIT (70/15/15)

X = df_final[feature_cols]
y = df_final['target'].astype(int)

n = len(X)
train_size = int(n * 0.7)
val_size   = int(n * 0.15)

X_train, y_train = X.iloc[:train_size], y.iloc[:train_size]
X_val, y_val     = X.iloc[train_size:train_size + val_size], y.iloc[train_size:train_size + val_size]
X_test, y_test   = X.iloc[train_size + val_size:], y.iloc[train_size + val_size:]

print(f"\nTrain shape : {X_train.shape} ({y_train.sum()} failure samples)")
print(f"Val shape   : {X_val.shape} ({y_val.sum()} failure samples)")
print(f"Test shape  : {X_test.shape} ({y_test.sum()} failure samples)")


# TRAINING XGBOOST

scale_pos_weight_moderat = np.sqrt((y_train == 0).sum() / max(1, (y_train == 1).sum()))
print(f"\nscale_pos_weight = {scale_pos_weight_moderat:.2f}")

model_full = xgb.XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight_moderat,
    eval_metric='aucpr',
    random_state=42,
    n_jobs=-1
)

model_full.fit(X_train, y_train)

# Validation
y_val_proba = model_full.predict_proba(X_val)[:, 1]

precisions, recalls, thresholds = precision_recall_curve(y_val, y_val_proba)
f1_scores = 2 * (precisions * recalls) / (precisions + recalls + 1e-10)
best_idx = np.argmax(f1_scores)
best_threshold = thresholds[best_idx] if best_idx < len(thresholds) else 0.5

print(f"\n- Threshold Tuning (Val Set) -")
print(f"Optimal Threshold : {best_threshold:.4f}")
print(f"Best Val F1-Score : {f1_scores[best_idx]:.4f}")

# Testing
y_test_proba = model_full.predict_proba(X_test)[:, 1]
y_test_pred_adj = (y_test_proba >= best_threshold).astype(int)

print("\n- Final Report Strict Early Warning System (Test Set) -")
print(classification_report(y_test, y_test_pred_adj, target_names=['Normal', 'Gagal']))
print(f"PR-AUC (Test): {average_precision_score(y_test, y_test_proba):.4f}")


# FEATURE IMPORTANCE ANALYSIS

importance_df = pd.DataFrame({
    'feature': feature_cols,
    'importance': model_full.feature_importances_
}).sort_values('importance', ascending=False).reset_index(drop=True)

print("\n- TOP 10 FEATURE IMPORTANCE -")
print(importance_df.head(10))

plt.figure(figsize=(10, 6))
plt.barh(importance_df['feature'].head(15)[::-1], importance_df['importance'].head(15)[::-1])
plt.xlabel('Importance Score')
plt.title('Top 15 Most Important Features')
plt.tight_layout()
plt.show()
```

###**COMPARISON**

```python
import numpy as np
import pandas as pd
import xgboost as xgb
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (
    classification_report, average_precision_score, precision_recall_curve,
    accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
)


# 1. FEATURE ENGINEERING


telemetry_df['datetime'] = pd.to_datetime(telemetry_df['datetime'])
errors_df['datetime']    = pd.to_datetime(errors_df['datetime'])
maint_df['datetime']     = pd.to_datetime(maint_df['datetime'])
failures_df['datetime']  = pd.to_datetime(failures_df['datetime'])

features = ['volt', 'rotate', 'pressure', 'vibration']

# Rolling 24h Mean & Std
telemetry_mean_24h = (telemetry_df.groupby('machineID')
                        .rolling(window=24, on='datetime')[features]
                        .mean().reset_index())
telemetry_mean_24h[features] = telemetry_mean_24h.groupby('machineID')[features].shift(1)

telemetry_sd_24h = (telemetry_df.groupby('machineID')
                      .rolling(window=24, on='datetime')[features]
                      .std().reset_index())
telemetry_sd_24h[features] = telemetry_sd_24h.groupby('machineID')[features].shift(1)

telemetry_mean_24h.columns = ['machineID', 'datetime'] + [f'{f}_mean_24h' for f in features]
telemetry_sd_24h.columns   = ['machineID', 'datetime'] + [f'{f}_sd_24h' for f in features]
telemetry_feat = pd.merge(telemetry_mean_24h, telemetry_sd_24h, on=['machineID', 'datetime'])

# Rolling 24h Min & Max
telemetry_min_24h = (telemetry_df.groupby('machineID')
                      .rolling(window=24, on='datetime')[features]
                      .min().reset_index())
telemetry_min_24h[features] = telemetry_min_24h.groupby('machineID')[features].shift(1)

telemetry_max_24h = (telemetry_df.groupby('machineID')
                      .rolling(window=24, on='datetime')[features]
                      .max().reset_index())
telemetry_max_24h[features] = telemetry_max_24h.groupby('machineID')[features].shift(1)

telemetry_min_24h.columns = ['machineID', 'datetime'] + [f'{f}_min_24h' for f in features]
telemetry_max_24h.columns = ['machineID', 'datetime'] + [f'{f}_max_24h' for f in features]

# Rate of Change
telemetry_diff = telemetry_df.copy()
for f in features:
    telemetry_diff[f'{f}_diff'] = telemetry_df.groupby('machineID')[f].diff()

tele_full = telemetry_feat.merge(telemetry_min_24h, on=['machineID', 'datetime'], how='left')
tele_full = tele_full.merge(telemetry_max_24h, on=['machineID', 'datetime'], how='left')
tele_full = tele_full.merge(
    telemetry_diff[['machineID', 'datetime', 'volt_diff', 'rotate_diff', 'pressure_diff', 'vibration_diff']],
    on=['machineID', 'datetime'], how='left'
)

# Error Rolling Count (24h & 72h)
error_dummies = pd.get_dummies(errors_df, columns=['errorID'])
error_cols = [c for c in error_dummies.columns if c.startswith('errorID_')]

error_timeline = pd.merge(telemetry_df[['datetime', 'machineID']], error_dummies, on=['datetime', 'machineID'], how='left').fillna(0)

error_count_24h = (error_timeline.groupby('machineID')
                   .rolling(window=24, on='datetime')[error_cols]
                   .sum().reset_index())
error_count_24h[error_cols] = error_count_24h.groupby('machineID')[error_cols].shift(1)
error_count_24h.columns = ['machineID', 'datetime'] + [f'{c}_count_24h' for c in error_cols]

error_count_72h = (error_timeline.groupby('machineID')
                   .rolling(window=72, on='datetime')[error_cols]
                   .sum().reset_index())
error_count_72h[error_cols] = error_count_72h.groupby('machineID')[error_cols].shift(1)
error_count_72h.columns = ['machineID', 'datetime'] + [f'{c}_count_72h' for c in error_cols]

error_feat = pd.merge(error_count_24h, error_count_72h, on=['machineID', 'datetime'])

# Maintenance History
maint_dummies = pd.get_dummies(maint_df, columns=['comp'])
comp_cols = [c for c in maint_dummies.columns if c.startswith('comp_')]

comp_df = pd.merge(telemetry_df[['datetime', 'machineID']], maint_dummies, on=['datetime', 'machineID'], how='left')

for c in comp_cols:
    comp_df[f'{c}_last'] = np.where(comp_df[c] == 1, comp_df['datetime'], pd.NaT)
    comp_df[f'{c}_last'] = comp_df.groupby('machineID')[f'{c}_last'].shift(1)
    comp_df[f'{c}_last'] = comp_df.groupby('machineID')[f'{c}_last'].ffill()
    comp_df[f'{c}_last'] = pd.to_datetime(comp_df[f'{c}_last'])
    comp_df[f'hours_since_{c}'] = (comp_df['datetime'] - comp_df[f'{c}_last']).dt.total_seconds() / 3600.0
    comp_df[f'hours_since_{c}'] = comp_df[f'hours_since_{c}'].fillna(1000)

maint_feat_cols = [f'hours_since_{c}' for c in comp_cols]
maint_feat = comp_df[['machineID', 'datetime'] + maint_feat_cols]

# Merge All Features & Metadata
machines_encoded = pd.get_dummies(machines_df, columns=['model'])

df_final = tele_full.merge(error_feat, on=['machineID', 'datetime'], how='left')
df_final = df_final.merge(maint_feat, on=['machineID', 'datetime'], how='left')
df_final = df_final.merge(machines_encoded, on='machineID', how='left')


# 2. TARGET LABELING & DATA SPLITTING


failures_df['failure_event'] = 1
df_final = pd.merge(df_final, failures_df[['datetime', 'machineID', 'failure_event']], on=['datetime', 'machineID'], how='left')
df_final['failure_event'] = df_final['failure_event'].fillna(0)

df_final = df_final.sort_values(by=['machineID', 'datetime']).reset_index(drop=True)

df_final['future_failure'] = df_final.groupby('machineID')['failure_event'].shift(-1)

indexer = pd.api.indexers.FixedForwardWindowIndexer(window_size=24)
df_final['target'] = (df_final.groupby('machineID')['future_failure']
                      .rolling(window=indexer, min_periods=1)
                      .max()
                      .reset_index(drop=True))

df_final['target'] = df_final['target'].fillna(0)
df_final = df_final.sort_values(by='datetime').reset_index(drop=True)

feature_cols = [c for c in df_final.columns if c not in ['datetime', 'machineID', 'failure_event', 'future_failure', 'target']]
df_final = df_final.dropna(subset=feature_cols)

X = df_final[feature_cols]
y = df_final['target'].astype(int)

n = len(X)
train_size = int(n * 0.7)
val_size   = int(n * 0.15)

X_train, y_train = X.iloc[:train_size], y.iloc[:train_size]
X_val, y_val     = X.iloc[train_size:train_size + val_size], y.iloc[train_size:train_size + val_size]
X_test, y_test   = X.iloc[train_size + val_size:], y.iloc[train_size + val_size:]


# 3. MODEL TRAINING & PREDICTION


# - A. Naive Baseline -
dummy_model = DummyClassifier(strategy='most_frequent', random_state=42)
dummy_model.fit(X_train, y_train)

y_test_pred_dummy = dummy_model.predict(X_test)
y_test_proba_dummy = dummy_model.predict_proba(X_test)[:, 1]

# - B. Logistic Regression Baseline -
baseline_lr = LogisticRegression(class_weight='balanced', max_iter=1000, random_state=42)
baseline_lr.fit(X_train, y_train)

y_test_pred_lr = baseline_lr.predict(X_test)
y_test_proba_lr = baseline_lr.predict_proba(X_test)[:, 1]

# - C. XGBoost (Early Warning System) -
scale_pos_weight_moderat = np.sqrt((y_train == 0).sum() / max(1, (y_train == 1).sum()))

model_full = xgb.XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight_moderat,
    eval_metric='aucpr',
    random_state=42,
    n_jobs=-1
)
model_full.fit(X_train, y_train)

# Threshold Tuning (Validation Set)
y_val_proba = model_full.predict_proba(X_val)[:, 1]
precisions, recalls, thresholds = precision_recall_curve(y_val, y_val_proba)
f1_scores = 2 * (precisions * recalls) / (precisions + recalls + 1e-10)
best_idx = np.argmax(f1_scores)
best_threshold = thresholds[best_idx] if best_idx < len(thresholds) else 0.5

# Inference Test Set with Best Threshold
y_test_proba_xgb = model_full.predict_proba(X_test)[:, 1]
y_test_pred_xgb = (y_test_proba_xgb >= best_threshold).astype(int)


# 4. Table Comparision Models


models = [
    ("Naive Baseline", y_test_pred_dummy, y_test_proba_dummy),
    ("Logistic Regression", y_test_pred_lr, y_test_proba_lr),
    ("XGBoost (EWS)", y_test_pred_xgb, y_test_proba_xgb)
]

comparison_results = []

for name, y_pred, y_proba in models:
    comparison_results.append({
        'Model': name,
        'Accuracy': round(accuracy_score(y_test, y_pred), 4),
        'Precision': round(precision_score(y_test, y_pred, pos_label=1, zero_division=0), 4),
        'Recall': round(recall_score(y_test, y_pred, pos_label=1, zero_division=0), 4),
        'F1-Score': round(f1_score(y_test, y_pred, pos_label=1, zero_division=0), 4),
        'PR-AUC': round(average_precision_score(y_test, y_proba), 4),
        'ROC-AUC': round(roc_auc_score(y_test, y_proba), 4)
    })

df_comp = pd.DataFrame(comparison_results)


print("\nTABEL PERBANDINGAN MODEL (TEST SET)\n")
print(df_comp.to_markdown(index=False))


plt.figure(figsize=(8, 5))
for name, y_pred, y_proba in models:
    precision_curve, recall_curve, _ = precision_recall_curve(y_test, y_proba)
    pr_auc = average_precision_score(y_test, y_proba)
    plt.plot(recall_curve, precision_curve, label=f"{name} (PR-AUC = {pr_auc:.4f})", linewidth=2)

plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Curve Comparison (Test Set)', fontweight='bold')
plt.legend(loc='lower left')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

###**TO VERIFY DATA LEAKAGE**

```python
import numpy as np
import pandas as pd
import xgboost as xgb
from sklearn.metrics import average_precision_score, classification_report



#  1: To test The extent of the model's performance depends on the errorID_* features.
print("="*70)
print("CHECK 1: ABLATION TEST -- the effect of errorID_* features on performance")
print("="*70)

error_cols_in_X = [c for c in X_train.columns if c.startswith('errorID_')]
print(f"Number of error features to be dropped: {len(error_cols_in_X)}")
print(f"Example feature: {error_cols_in_X[:5]}")

X_train_no_err = X_train.drop(columns=error_cols_in_X)
X_val_no_err   = X_val.drop(columns=error_cols_in_X)
X_test_no_err  = X_test.drop(columns=error_cols_in_X)

scale_pos_weight_moderat = np.sqrt((y_train == 0).sum() / max(1, (y_train == 1).sum()))

model_no_err = xgb.XGBClassifier(
    n_estimators=300, max_depth=6, learning_rate=0.05,
    subsample=0.8, colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight_moderat,
    eval_metric='aucpr', random_state=42, n_jobs=-1
)
model_no_err.fit(X_train_no_err, y_train)
y_test_proba_no_err = model_no_err.predict_proba(X_test_no_err)[:, 1]

pr_auc_with_err = average_precision_score(y_test, y_test_proba)
pr_auc_no_err = average_precision_score(y_test, y_test_proba_no_err)

print(f"\nPR-AUC with error feature : {pr_auc_with_err:.4f}")
print(f"PR-AUC without error feature : {pr_auc_no_err:.4f}")
print(f"PR-AUC Difference            : {pr_auc_with_err - pr_auc_no_err:.4f}")

if pr_auc_no_err < 0.6:
    print(">> Performance is heavily dependent on error features. This requires further investigation:")
elif pr_auc_no_err < 0.85:
    print(">> Performance relies significantly on error features, but the model still retains a signal.")

else:
    print(">> The model remains robust without error features—performance is not dominated by a single feature group..")



# 2. checking whether the recall we obtained actually means the model is capable of detecting a wide range of different failure events.

print("\n" + "="*70)
print("CEK 2: EVENT-LEVEL DETECTION RATE")
print("="*70)



test_meta = df_final.loc[X_test.index, ['machineID', 'datetime', 'future_failure']].copy()
test_meta['y_true'] = y_test.values
test_meta['y_pred'] = (y_test_proba >= best_threshold).astype(int)

# Actual failure events: rows where future failure == 1 (failure occurs EXACTLY at t+1 or 1 hour after)
actual_failure_points = test_meta[test_meta['future_failure'] == 1][['machineID', 'datetime']].copy()
actual_failure_points = actual_failure_points.rename(columns={'datetime': 'failure_time'})

print(f"Number of UNIQUE failure events in the test set: {len(actual_failure_points)}")

detected = 0
for _, row in actual_failure_points.iterrows():
    m_id = row['machineID']
    f_time = row['failure_time']
    # Detection window: 24 hours before failure_time (rows labeled 1 for this event)
    window_start = f_time - pd.Timedelta(hours=23)
    mask = (
        (test_meta['machineID'] == m_id) &
        (test_meta['datetime'] >= window_start) &
        (test_meta['datetime'] <= f_time) &
        (test_meta['y_pred'] == 1)
    )
    if mask.any():
        detected += 1

event_level_recall = detected / len(actual_failure_points) if len(actual_failure_points) > 0 else 0
row_level_recall_reported = 0.96

print(f"\nEvent-level detection rate : {detected}/{len(actual_failure_points)} = {event_level_recall:.4f}")
print(f"Row-level recall (previously reported): {row_level_recall_reported:.4f}")
print(f"Recall Difference : {row_level_recall_reported - event_level_recall:.4f}")

if event_level_recall < row_level_recall_reported - 0.15:
    print(">> Event-level recall is FAR lower than row-level recall.")
    print("   Row-level recall has consistently overestimated the model's actual performance.")

else:
    print(">> Event-level recall is quite close to row-level recall.")
    print("   Row-level metrics do indeed reasonably reflect actual detection capabilities.")


# CHECK 3: Correlation of each error feature with the target

print("\n" + "="*70)
print("CEK 3: Correlation of Error Features with the Target")
print("="*70)
print("A very high individual correlation (>0.7–0.8) is a common red flag for leakage.\n")

corr_with_target = X_train[error_cols_in_X].apply(lambda col: col.corr(y_train)).sort_values(key=abs, ascending=False)
print(corr_with_target.head(10))

max_corr = corr_with_target.abs().max()
if max_corr > 0.7:
    print(f"\n>> The highest correlation ({max_corr:.3f}) is classified as VERY HIGH for an individual feature.")

else:
    print(f"\n>> The highest correlation ({max_corr:.3f}) is still within a reasonable range.")


# SUMMARY

print("\n" + "="*70)
print("SUMMARY")
print("="*70)
print(f"PR-AUC with error features       : {pr_auc_with_err:.4f}")
print(f"PR-AUC wihtout error features        : {pr_auc_no_err:.4f}")
print(f"Row-level recall (Failure class) : {row_level_recall_reported:.4f}")
print(f"Event-level detection rate       : {event_level_recall:.4f}")
print(f"Maximum correlation of error features : {max_corr:.4f}")
```

###**TIME ANALYSIS**

```python
import pandas as pd
import numpy as np



lead_times = []
never_detected_early = []

for _, row in actual_failure_points.iterrows():
    m_id = row['machineID']
    f_time = row['failure_time']
    window_start = f_time - pd.Timedelta(hours=23)

    window_rows = test_meta[
        (test_meta['machineID'] == m_id) &
        (test_meta['datetime'] >= window_start) &
        (test_meta['datetime'] <= f_time)
    ].sort_values('datetime')

    positive_rows = window_rows[window_rows['y_pred'] == 1]

    if len(positive_rows) > 0:
        first_alarm_time = positive_rows['datetime'].iloc[0]
        lead_time_hours = (f_time - first_alarm_time).total_seconds() / 3600.0
        lead_times.append(lead_time_hours)
    else:
        never_detected_early.append((m_id, f_time))

lead_times = np.array(lead_times)

print("="*70)
print("CHECK 4: LEAD TIME ANALYSIS (warning hours before failure)")
print("="*70)
print(f"Number of events with at least 1 alarm: {len(lead_times)}")
print(f"Number of events WITHOUT any alarm: {len(never_detected_early)}")

if len(lead_times) > 0:
    print(f"\nLead time statistics (hours before failure when the FIRST alarm appears):")
    print(f"  Average   : {lead_times.mean():.2f} hours")
    print(f"  Median    : {np.median(lead_times):.2f} hours")
    print(f"  Minimum   : {lead_times.min():.2f} hours")
    print(f"  Maximum   : {lead_times.max():.2f} hours")

    print(f"\nLead time distribution:")
    print(f"  >= 20 hours (very early warning)      : {(lead_times >= 20).sum()} events ({(lead_times >= 20).mean()*100:.1f}%) ")
    print(f"  10-19 hours (early enough warning)    : {((lead_times >= 10) & (lead_times < 20)).sum()} events ({(((lead_times >= 10) & (lead_times < 20)).mean())*100:.1f}%) ")
    print(f"  1-9 hours (close warning)             : {((lead_times >= 1) & (lead_times < 10)).sum()} events ({(((lead_times >= 1) & (lead_times < 10)).mean())*100:.1f}%) ")
    print(f"  < 1 hour (almost no reaction time)    : {(lead_times < 1).sum()} events ({(lead_times < 1).mean()*100:.1f}%) ")

    if lead_times.mean() < 5:
        print("\n>> The average lead time is quite short. Although the model detects")
        print(" 100% of failure events, it may detect them only shortly before the failure happens.")
        print(" This means there may not be enough time to take preventive action.")
        print(" Therefore, the model's value as an early warning system should be reconsidered.")



    else:
        print(f"\n>> The average lead time of {lead_times.mean():.1f} hours is considered reasonable for an early")
        print("   warning system that is practically useful.")
```
